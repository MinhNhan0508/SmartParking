# License plate recognition demo

Run the same two-stage YOLO pipeline used by the desktop terminals. Install
`requirements-dev.txt` first. `PARKING_TEST_IMAGE` selects a local test image;
the default is a report screenshot, for demonstration only, not evaluation.

```python
from pathlib import Path
import os
import sys
import cv2
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "license_plate_module.py").is_file():
    ROOT = ROOT.parent
if not (ROOT / "license_plate_module.py").is_file():
    raise RuntimeError("Open this notebook from the SmartParking repository.")
sys.path.insert(0, str(ROOT))
from license_plate_module import recognize_plate_from_image
```

```python
TEST_IMAGE_PATH = Path(os.getenv("PARKING_TEST_IMAGE", str(ROOT / "docs/images/check-in.png")))
if not TEST_IMAGE_PATH.is_file():
    raise FileNotFoundError(f"Choose an existing image with PARKING_TEST_IMAGE: {TEST_IMAGE_PATH}")

crop, plate_text = recognize_plate_from_image(str(TEST_IMAGE_PATH))
print("Recognized plate:", plate_text)
if crop is not None:
    plt.figure(figsize=(8, 4))
    plt.imshow(cv2.cvtColor(crop, cv2.COLOR_BGR2RGB))
    plt.title(plate_text)
    plt.axis("off")
    plt.show()
```

This notebook does not train YOLO or measure dataset accuracy. The original
plate training dataset is not bundled. See the project report for the original
demonstration and the README for limitations.
